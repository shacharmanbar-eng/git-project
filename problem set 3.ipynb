```python
def print_message(message, n):
    for i in range(n):
        print(message)

print_message("E.T. come home", 5)
```

```python
def to_question(text):
    if not text.endswith("?"):
        return text + "?"
    return text

print(to_question("How do you do"))   # מה שלומך?
print(to_question("How do you do?"))  # מה שלומך?
```

```python
def text_analyze(text):
    words_count = len(text.split())
    chars_count = len(text.replace(" ", ""))
    return words_count, chars_count

print(text_analyze("טוב למות בעד ארצנו"))
```

```python
def palindrome_is(text):
    text = text.replace(" ", "").lower()
    return text == text[::-1]

print(palindrome_is("Never odd or even"))  # True
print(palindrome_is("Hello"))              # False
```

```python
def frequency_word(text):
    frequencies = {}

    for word in text.split():
        frequencies[word] = frequencies.get(word, 0) + 1

    return frequencies

print(frequency_word("גנן גידל דגן בגן דגן גדול גדל בגן"))
```

```python
def word_longest(text):
    words = text.split()
    return max(words, key=len, default="")

print(word_longest("נ נח נחמ נחמן מאומן"))  
```

```python
def length_word_average(text):
    words = text.split()

    if not words:
        return 0.0

    average = sum(len(word) for word in words) / len(words)
    return round(average, 2)

print(length_word_average("תמיד הכי חשוך לפני הזריחה"))  # 3.5
```
